```python
# ------------------dictionary ve tuple key value olma mevzusu (kafa karıştırıcıydı baya) ------------------
b = {}
for w in words[:2]:
    chs = ['<S>'] + list(w) + ['<E>']
    for ch1, ch2 in zip(chs, chs[1:]):
        biagram = (ch1, ch2)  #tuplemış bunun adı
        # b.append(biagram)  # bu append hatası veriyor çünkü b bir dict, dict'lerde append yok.
        b[biagram] = b.get(biagram, 0) + 1   #dictin key valuesı biagram burda, biagramı koyup get yapıyoruz ve onun değerini buluyoruz, eğer dictte yoksa 0 oluyor valuesı(parantezdeki ikinci arguman), eger varsa +1 ekliyoruz zaten.
        print(biagram, b[biagram])    
```

```python
# ------------------sorted() mevzusu------------------
# sorted(b.items(), key = lambda kv: -kv[1])
sorted(b.items(), key = lambda kv: -kv[1]) # b.items() dictin key value çiftlerini döndürüyor, key = lambda kv: -kv[1] ise bu çiftlerin value'suna göre sıralama yapıyor, - işareti ile azalan sırada sıralıyor.
```

```python
# ------------------dcitionary içinde s ve i yi ters çevirip yeni bir dict yapma işi------------------

chars = sorted(list(set(''.join(words))))

stoi = {s:i+1 for i,s in enumerate(chars)} #s:i+1 burda aslında enum 0 dan başlatmak yerıne 1 den  başlatıyor ve sayı-harf yeri,ne harf-sayı yapıyor işi
stoi['.'] = 0
itos = {i:s for s,i in stoi.items()} #burada ıtems yazmazsak sadece key valuelar dönüyormuş, onu da bölemediği için sıkıntı oluyor. items() yazınca ikili değerler dönüyor key ve value
itos
```
