```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
words = open('names.txt', 'r').read().splitlines()
words[:5]
```

```python
len(words)
```

```python
chars = sorted(list(set(''.join(words))))
stoi = {s:i+1 for i, s in enumerate(chars)}
stoi['.'] = 0
itos = {i:s for s, i in stoi.items()} #sadece stoi yazmak kv döndürür, stoi.items() yazarsan kv, value ciftlerini döndürür (tuple formatinda)
itos
```

```python
#dataset

block_size = 3
X, Y = [], []
for w in words[:5]:

    print(w)
    context = [0] * block_size
    for ch in w + '.':
        ix = stoi[ch]
        X.append(context)
        Y.append(ix)
        print(''.join(itos[i] for i in context), '--->', itos[ix])
        context = context[1:] + [ix] #contexti kaydırıp yeni karakteri ekliyoruz
```
